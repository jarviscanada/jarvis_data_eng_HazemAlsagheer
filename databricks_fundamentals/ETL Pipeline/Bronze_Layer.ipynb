# Data Ingestion

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Bronze Layer").getOrCreate()
```

```python
url = "jdbc:sqlserver://jrvs-server.database.windows.net:1433;database=jrvs-database;user=jarvis_server@jrvs-server;password=1234Jrvs.;encrypt=true;trustServerCertificate=false;hostNameInCertificate=*.database.windows.net;loginTimeout=30;"
 
raw_transactions_df = (spark.read
  .format("jdbc")
  .option("url", url)
  .option("dbtable", "transactions_data")
  .option("user", "jarvis_server")
  .option("password", "1234Jrvs.")
  .option("driver", "com.microsoft.sqlserver.jdbc.SQLServerDriver")
  .load()
)
raw_transactions_df.write.mode('overwrite').saveAsTable('databricks_fundamentals.bronze.transactions_data')

```
